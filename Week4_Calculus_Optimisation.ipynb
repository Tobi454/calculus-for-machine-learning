## 1. Functions

A function describes a relationship between an input and an output. It takes one or more inputs and produces an output according to a rule.

For example:

\[
f(x)=x^2+2x+1
\]

If \(x=2\), then:

\[
f(2)=2^2+2(2)+1=9
\]

The input is 2, and the output is 9.

### Functions in Machine Learning

Functions are used throughout Machine Learning. A model can be viewed as a function that takes input data and produces a prediction.

For example, a simple linear regression model is:

\[
\hat{y}=wx+b
\]

Where:

- \(x\) is the input feature.
- \(w\) is the weight.
- \(b\) is the bias.
- \(\hat{y}\) is the predicted output.

Suppose a model predicts a house price using its size:

\[
\hat{y}=200x+50000
\]

If \(x=100\), then:

\[
\hat{y}=200(100)+50000=70000
\]

The model predicts a price of 70,000 in the chosen currency units.

The goal during training is to find suitable values of \(w\) and \(b\) so that the predictions are close to the actual values.

## 2. Limits

A limit describes the value that a function approaches as its input approaches a particular number.

Limits are important because they help define derivatives and describe the behaviour of functions near specific points.

The notation is:

\[
\lim_{x\to a}f(x)=L
\]

This means that as \(x\) approaches \(a\), the function \(f(x)\) approaches \(L\).

### Example 1: Direct substitution

Find:

\[
\lim_{x\to 2}(x^2+3)
\]

Substitute \(x=2\):

\[
2^2+3=7
\]

Therefore:

\[
\boxed{\lim_{x\to 2}(x^2+3)=7}
\]

### Example 2: Simplifying before substitution

Find:

\[
\lim_{x\to 2}\frac{x^2-4}{x-2}
\]

Direct substitution gives \(0/0\), which is an indeterminate form.

Factor the numerator:

\[
x^2-4=(x-2)(x+2)
\]

For \(x\ne 2\), simplify:

\[
\frac{(x-2)(x+2)}{x-2}=x+2
\]

Now take the limit:

\[
\lim_{x\to 2}(x+2)=4
\]

Therefore:

\[
\boxed{4}
\]

The function is undefined at \(x=2\), but its limit as \(x\) approaches 2 is still 4.

### Why limits matter in Machine Learning

Limits help establish the mathematical definition of a derivative. Derivatives describe how a function changes, and they are fundamental to optimization algorithms used to train many Machine Learning models.

## 3. Derivatives and Differentiation

A derivative measures the instantaneous rate of change of a function with respect to its input.

It also represents the slope of the tangent line to a function's graph at a particular point.

For a function \(f(x)\), its derivative can be written as:

\[
f'(x),\quad \frac{dy}{dx},\quad \text{or}\quad \frac{df}{dx}
\]

### 3.1 Derivative from first principles

The derivative is defined using a limit:

\[
f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
\]

This formula measures the change in the function divided by the change in its input as the input difference approaches zero.

Consider:

\[
f(x)=x^2
\]

Using first principles:

\[
f'(x)=\lim_{h\to 0}\frac{(x+h)^2-x^2}{h}
\]

Expand the expression:

\[
f'(x)=\lim_{h\to 0}\frac{x^2+2xh+h^2-x^2}{h}
\]

Simplify:

\[
f'(x)=\lim_{h\to 0}\frac{2xh+h^2}{h}
\]

Factor out \(h\) and cancel:

\[
f'(x)=\lim_{h\to 0}(2x+h)
\]

Taking the limit gives:

\[
\boxed{f'(x)=2x}
\]

This means that the rate of change of \(x^2\) at any point \(x\) is \(2x\).

For example, at \(x=3\):

\[
f'(3)=2(3)=6
\]

The slope of the tangent to the curve at \(x=3\) is 6.

### 3.2 Basic differentiation rules

#### Constant rule

The derivative of a constant is zero.

\[
\frac{d}{dx}(c)=0
\]

Example:

\[
\frac{d}{dx}(8)=0
\]

A constant does not change when \(x\) changes.

#### Power rule

For \(f(x)=x^n\):

\[
\frac{d}{dx}(x^n)=nx^{n-1}
\]

Examples:

\[
\frac{d}{dx}(x^3)=3x^2
\]

\[
\frac{d}{dx}(x^5)=5x^4
\]

\[
\frac{d}{dx}(x)=1
\]

For the last example, \(x=x^1\), so its derivative is \(1x^0=1\).

#### Constant multiple rule

A constant multiplier remains outside the derivative.

\[
\frac{d}{dx}[cf(x)]=c\frac{d}{dx}f(x)
\]

Example:

\[
\frac{d}{dx}(4x^3)=12x^2
\]

#### Sum and difference rules

Differentiate each term separately.

\[
\frac{d}{dx}[f(x)+g(x)]=f'(x)+g'(x)
\]

Example:

\[
f(x)=3x^3+2x^2-5x+7
\]

Differentiating:

\[
\boxed{f'(x)=9x^2+4x-5}
\]

The constant \(7\) disappears because its derivative is zero.

### Why derivatives matter in Machine Learning

Suppose a model has a loss function that measures prediction error. The derivative of the loss with respect to a parameter tells us how the loss changes when that parameter changes slightly.

- A positive derivative means a small increase in the parameter increases the loss locally.
- A negative derivative means a small increase in the parameter decreases the loss locally.
- A zero derivative means the function is locally flat to first order. It may be a minimum, a maximum, or another stationary point.

This information helps us decide how to update model parameters.

## 4. Product, Quotient, and Chain Rules

Some functions require special differentiation rules because they combine other functions through multiplication, division, or composition.

### 4.1 Product rule

Use the product rule when two differentiable functions are multiplied together.

If:

\[
y=u(x)v(x)
\]

Then:

\[
\boxed{\frac{dy}{dx}=u'v+uv'}
\]

In words, differentiate the first function and multiply it by the second, then add the first function multiplied by the derivative of the second.

Example:

\[
y=x^2(x+3)
\]

Let:

\[
u=x^2,\qquad v=x+3
\]

Then:

\[
u'=2x,\qquad v'=1
\]

Applying the product rule:

\[
\frac{dy}{dx}=2x(x+3)+x^2(1)
\]

Simplifying:

\[
\boxed{\frac{dy}{dx}=3x^2+6x}
\]

Expanding the original function first and then differentiating would produce the same result.

### 4.2 Quotient rule

Use the quotient rule when one differentiable function is divided by another.

If:

\[
y=\frac{u(x)}{v(x)},\qquad v(x)\ne 0
\]

Then:

\[
\boxed{\frac{dy}{dx}=\frac{vu'-uv'}{v^2}}
\]

Example:

\[
y=\frac{x^2+1}{x}
\]

Let:

\[
u=x^2+1,\qquad v=x
\]

Then:

\[
u'=2x,\qquad v'=1
\]

Applying the quotient rule:

\[
\frac{dy}{dx}=\frac{x(2x)-(x^2+1)(1)}{x^2}
\]

Simplifying:

\[
\boxed{\frac{dy}{dx}=\frac{x^2-1}{x^2}}
\]

This result applies for \(x\ne 0\), where the original function is defined.

### 4.3 Chain rule

The chain rule is used when one function is inside another function.

If:

\[
y=f(g(x))
\]

Then:

\[
\boxed{\frac{dy}{dx}=f'(g(x))g'(x)}
\]

In simple terms, differentiate the outer function while keeping the inner function in place, then multiply by the derivative of the inner function.

Example:

\[
y=(3x+2)^4
\]

The outer function raises a value to the fourth power, while the inner function is \(3x+2\).

Differentiate the outer function:

\[
4(3x+2)^3
\]

Differentiate the inner function:

\[
\frac{d}{dx}(3x+2)=3
\]

Multiply the results:

\[
\boxed{\frac{dy}{dx}=12(3x+2)^3}
\]

### 4.4 Chain rule in Machine Learning

The chain rule is especially important in neural networks, where one operation feeds into another.

For example, consider a simple model:

\[
\hat{y}=wx
\]

Suppose its squared-error loss for one observation is:

\[
L(w)=(wx-y)^2
\]

We want to differentiate the loss with respect to the weight \(w\).

Let the inner expression be:

\[
u=wx-y
\]

Then:

\[
L=u^2
\]

Differentiating the outer function gives:

\[
\frac{dL}{du}=2u
\]

Differentiating the inner expression with respect to \(w\) gives:

\[
\frac{du}{dw}=x
\]

Using the chain rule:

\[
\frac{dL}{dw}=\frac{dL}{du}\frac{du}{dw}
\]

Therefore:

\[
\boxed{\frac{dL}{dw}=2(wx-y)x}
\]

This derivative tells us how the squared-error loss changes as the weight changes. Training algorithms use this kind of calculation to update parameters.

In neural networks, the repeated application of the chain rule across layers is a central part of **backpropagation**.

## 5. Partial Derivatives

A function may depend on more than one variable. A partial derivative measures how the function changes with respect to one variable while the other variables are held constant.

For a function:

\[
f(x,y)=x^2+3y^2
\]

The partial derivative with respect to \(x\) is written as:

\[
\frac{\partial f}{\partial x}
\]

The partial derivative with respect to \(y\) is written as:

\[
\frac{\partial f}{\partial y}
\]

The symbol \(\partial\) is used instead of \(d\) because the function depends on multiple variables.

### 5.1 Partial derivative with respect to \(x\)

Start with:

\[
f(x,y)=x^2+3y^2
\]

When differentiating with respect to \(x\), treat \(y\) as a constant.

Therefore:

\[
\boxed{\frac{\partial f}{\partial x}=2x}
\]

The term \(3y^2\) has derivative zero with respect to \(x\), since it does not change as \(x\) changes while \(y\) is held fixed.

### 5.2 Partial derivative with respect to \(y\)

When differentiating with respect to \(y\), treat \(x\) as a constant.

Therefore:

\[
\boxed{\frac{\partial f}{\partial y}=6y}
\]

### 5.3 Evaluating partial derivatives at a point

Suppose:

\[
f(x,y)=x^2+3y^2
\]

At the point \((2,1)\):

\[
\frac{\partial f}{\partial x}=2(2)=4
\]

\[
\frac{\partial f}{\partial y}=6(1)=6
\]

The partial derivatives are 4 and 6 at that point.

The first value describes the rate of change in the \(x\)-direction, while the second describes the rate of change in the \(y\)-direction.

### 5.4 Partial derivatives in Machine Learning

Machine Learning models often have many parameters. A loss function may depend on multiple weights and biases.

For example:

\[
L(w_1,w_2)=(w_1x_1+w_2x_2-y)^2
\]

The loss depends on both \(w_1\) and \(w_2\).

We can calculate:

\[
\frac{\partial L}{\partial w_1}
\qquad\text{and}\qquad
\frac{\partial L}{\partial w_2}
\]

Each partial derivative tells us how the loss changes with respect to one parameter while the other is held constant.

This lets us update multiple parameters during training.

## 6. Gradients

The gradient extends the idea of a derivative to functions with multiple input variables. It is a vector containing all the first-order partial derivatives of a scalar-valued function.

For:

\[
f(x,y)
\]

the gradient is:

\[
\boxed{
\nabla f=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
}
\]

The symbol \(\nabla\) is called *nabla* or *del*.

### 6.1 Calculating a gradient

Consider:

\[
f(x,y)=x^2+3y^2
\]

We already found:

\[
\frac{\partial f}{\partial x}=2x
\]

\[
\frac{\partial f}{\partial y}=6y
\]

Therefore, the gradient is:

\[
\boxed{
\nabla f(x,y)=
\begin{bmatrix}
2x\\
6y
\end{bmatrix}
}
\]

At the point \((2,1)\):

\[
\nabla f(2,1)=
\begin{bmatrix}
4\\
6
\end{bmatrix}
\]

This vector describes how the function changes in its input space at that point.

### 6.2 What does the gradient tell us?

For a differentiable function, the gradient points in the direction of the steepest local increase, provided the gradient is not zero.

The negative gradient points in the direction of steepest local decrease.

This is why gradient-based optimization methods use the negative gradient to update parameters when minimizing a loss function.

If:

\[
\nabla L(\theta)
\]

is the gradient of the loss with respect to the model parameters \(\theta\), a basic gradient-descent update is:

\[
\boxed{\theta_{\text{new}}=\theta_{\text{old}}-\eta\nabla L(\theta_{\text{old}})}
\]

Where:

- \(\theta\) represents the model parameters.
- \(L\) is the loss function.
- \(\nabla L\) is the gradient of the loss.
- \(\eta\) is the learning rate.

The learning rate determines the size of each update.

### 6.3 Why gradients matter

Without gradients, an algorithm would not know which direction to adjust its parameters to reduce the loss efficiently.

Gradients provide a systematic way to determine how each parameter contributes to changes in the loss. They are used in gradient descent, neural network training, and many other optimization methods.

## 7. Optimization

Optimization is the process of finding the best values of variables or parameters according to a particular objective.

In Machine Learning, the objective is often to minimize a loss function that measures how far a model's predictions are from the target values.

For example, a regression model may use mean squared error:

\[
MSE=\frac{1}{n}\sum_{i=1}^{n}(\hat{y}_i-y_i)^2
\]

Where:

- \(n\) is the number of observations.
- \(\hat{y}_i\) is a prediction.
- \(y_i\) is the actual value.

A lower MSE generally indicates that predictions are closer to the observed values on the dataset used to calculate it.

The optimization problem is to find parameter values that minimize this loss.

### 7.1 Stationary points

A stationary point is a point where the first derivative is zero:

\[
f'(x)=0
\]

Stationary points may correspond to local minima, local maxima, or other points where the slope is zero.

To locate potential extrema, we can differentiate a function and solve for where the derivative equals zero.

### 7.2 Local and global minima

A **local minimum** is a point where the function's value is no greater than the values at nearby points.

A **global minimum** is a point where the function achieves its lowest value over the entire domain being considered.

For example:

\[
f(x)=x^2
\]

Its derivative is:

\[
f'(x)=2x
\]

Set the derivative equal to zero:

\[
2x=0
\]

Therefore:

\[
x=0
\]

Since \(x^2\geq 0\) for every real \(x\), the function has a global minimum at \(x=0\), with \(f(0)=0\).

### 7.3 Local and global maxima

A **local maximum** is a point where the function's value is no less than the values at nearby points.

A **global maximum** is a point where the function achieves its highest value over the domain being considered.

For example:

\[
f(x)=-x^2
\]

Its derivative is:

\[
f'(x)=-2x
\]

Setting the derivative to zero gives \(x=0\). Since the function is never greater than zero, it has a global maximum at \(x=0\).

### 7.4 The second derivative test

The second derivative can help classify a stationary point.

After finding a point \(x=a\) where \(f'(a)=0\):

- If \(f''(a)>0\), the function has a local minimum at \(a\).
- If \(f''(a)<0\), the function has a local maximum at \(a\).
- If \(f''(a)=0\), the test is inconclusive.

For example:

\[
f(x)=x^2
\]

First derivative:

\[
f'(x)=2x
\]

Second derivative:

\[
f''(x)=2
\]

At \(x=0\), the second derivative is positive, so \(x=0\) is a local minimum. It is also the global minimum.

The second derivative test does not always provide an answer, and a stationary point is not necessarily a minimum or maximum.

### 7.5 Optimization in Machine Learning

The loss function of a Machine Learning model depends on its parameters. Training attempts to find parameter values that produce a low loss.

For a simple model with one parameter \(w\), we might want to minimize:

\[
L(w)=w^2
\]

The minimum occurs at \(w=0\). Starting from another value, an optimization algorithm can repeatedly adjust \(w\) to move toward that minimum.

Real Machine Learning loss functions can have many parameters and complicated surfaces. In such cases, optimization algorithms use gradients and repeated updates instead of solving the entire problem analytically.

## 8. Gradient Descent

Gradient descent is an optimization algorithm that minimizes a function by repeatedly moving its parameters in the direction opposite to the gradient.

It is widely used to train Machine Learning models.

The update rule is:

\[
\boxed{w_{\text{new}}=w_{\text{old}}-\eta\frac{dL}{dw}}
\]

For a model with multiple parameters, the same idea becomes:

\[
\boxed{\theta_{\text{new}}=\theta_{\text{old}}-\eta\nabla L(\theta_{\text{old}})}
\]

Where:

- \(w\) or \(\theta\) represents the parameters.
- \(L\) is the loss function.
- \(\frac{dL}{dw}\) or \(\nabla L\) is the derivative or gradient.
- \(\eta\) is the learning rate.

### 8.1 Understanding the learning rate

The learning rate controls how far the parameters move during each update.

- A very small learning rate may make training slow.
- A very large learning rate may cause the algorithm to overshoot a minimum or fail to converge.
- A suitable learning rate helps the algorithm make useful progress toward a low-loss region.

The best learning rate depends on the loss function, parameter scales, and optimization method.

### 8.2 Step-by-step example

Consider the loss function:

\[
L(w)=w^2
\]

Its derivative is:

\[
\frac{dL}{dw}=2w
\]

Suppose we start with:

\[
w=4,\qquad \eta=0.1
\]

The initial loss is:

\[
L(4)=4^2=16
\]

Calculate the gradient:

\[
\frac{dL}{dw}=2(4)=8
\]

Apply the update rule:

\[
w_{\text{new}}=4-0.1(8)
\]

\[
\boxed{w_{\text{new}}=3.2}
\]

The new loss is:

\[
L(3.2)=3.2^2=10.24
\]

The loss has decreased from 16 to 10.24 after one update.

If we repeat this process, the parameter moves closer to zero, and the loss decreases toward zero.

### 8.3 Implementing gradient descent using Python

The following example uses NumPy to calculate the gradient and Matplotlib to plot the loss during optimization.

**Python code cell:**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the loss function
def loss(w):
    return w ** 2

# Define the derivative of the loss function
def gradient(w):
    return 2 * w

# Set the initial parameter and learning rate
w = 4.0
learning_rate = 0.1
iterations = 30

# Store the loss after each iteration
loss_history = []

for i in range(iterations):
    current_loss = loss(w)
    loss_history.append(current_loss)

    # Update the parameter using gradient descent
    w = w - learning_rate * gradient(w)

print("Final parameter:", w)
print("Final loss:", loss(w))

# Plot the loss over the iterations
plt.plot(range(iterations), loss_history)
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Gradient Descent")
plt.grid(True)
plt.show()
```

### What the code does

1. Defines the loss function \(L(w)=w^2\).
2. Defines its derivative, \(L'(w)=2w\).
3. Starts with \(w=4\).
4. Calculates the gradient at the current parameter.
5. Updates the parameter using the learning rate.
6. Repeats the process for 30 iterations.
7. Plots the recorded loss values.

The loss should decrease as the parameter approaches zero.

The final loss may not be exactly zero because the algorithm performs a finite number of updates.

### 8.4 What this example teaches

This example demonstrates the basic mechanics of gradient descent. It uses a simple function whose minimum is already known, so we can easily see whether the algorithm moves in the correct direction.

In a real Machine Learning problem, the loss function is calculated from a dataset, and the parameters may include many weights and biases. The same underlying principle applies, although practical training can involve additional considerations such as data batches, feature scaling, regularization, and more advanced optimizers.

## 9. Connecting Calculus to Machine Learning

The topics covered in this note work together during model training.

Consider a regression model that predicts house prices using features such as house size.

### Step 1: Define the model

A simple linear model can be written as:

\[
\hat{y}=wx+b
\]

The weight \(w\) and bias \(b\) determine the predictions.

### Step 2: Measure prediction error

A loss function measures how far predictions are from actual values. For one observation, a squared-error loss is:

\[
L=(\hat{y}-y)^2
\]

Substituting the model:

\[
L=(wx+b-y)^2
\]

### Step 3: Calculate derivatives

We differentiate the loss with respect to the parameters.

For the weight:

\[
\frac{\partial L}{\partial w}=2(wx+b-y)x
\]

For the bias:

\[
\frac{\partial L}{\partial b}=2(wx+b-y)
\]

These derivatives tell us how the loss changes with respect to each parameter.

### Step 4: Update the parameters

Gradient descent updates both parameters in the direction that locally reduces the loss:

\[
w_{\text{new}}=w_{\text{old}}-\eta\frac{\partial L}{\partial w}
\]

\[
b_{\text{new}}=b_{\text{old}}-\eta\frac{\partial L}{\partial b}
\]

### Step 5: Repeat

The model calculates predictions, evaluates the loss, calculates gradients, and updates its parameters repeatedly.

This process continues until a stopping condition is met, such as reaching a set number of iterations or seeing little improvement in the loss.

### How the concepts connect

- **Functions** describe the model's predictions and loss.
- **Limits** help define derivatives mathematically.
- **Derivatives** measure how a function changes.
- **The chain rule** helps calculate derivatives through multiple operations.
- **Partial derivatives** measure how a function changes with respect to individual parameters.
- **Gradients** collect these partial derivatives into a vector.
- **Optimization** seeks parameter values that minimize or maximize an objective.
- **Gradient descent** uses gradients to update parameters and reduce loss.

Understanding these ideas provides a mathematical foundation for studying linear regression, logistic regression, neural networks, and other gradient-based Machine Learning methods.
